```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
from skimage.util import random_noise
import matplotlib.pyplot as plt
```

```python
img=cv2.imread("/content/dog4.jpeg")
cv2_imshow(img)
```

```python
def median_blur(img):
  noise_img=random_noise(img,mode='s&p',amount=0.15)
  noise_img=np.array(255*noise_img,dtype='uint8')
  median=cv2.medianBlur(noise_img,5)
  return median,noise_img
```

```python
def min_filter(img):
  size=(3,3)
  shape=cv2.MORPH_RECT
  kernel=cv2.getStructuringElement(shape,size)
  min_image=cv2.erode(img,kernel)
  return min_image
```

```python
def max_filter(img):
  size=(3,3)
  shape=cv2.MORPH_RECT
  kernel=cv2.getStructuringElement(shape,size)
  max_image=cv2.dilate(img,kernel)
  return max_image
```

```python
def bilateral(img):
  bil=cv2.bilateralFilter(img,15,80,80)
  return bil
```

```python
def main():
  print("Different types of Non-Linear Spatial Filtering")
  while True:
     print('\n1.Median\n2.Minimum\n3.Maximum\n4.Bilateral\n5.Exit')
     ch=int(input('Enter your choice : '))
     if ch==1:
       print("\nNoise Image")
       md,noise_img=median_blur(img)
       cv2_imshow(noise_img)
       print("Denoising by Median Blur")
       cv2_imshow(md)
     elif ch==2:
       print("\nOriginal Image ")
       cv2_imshow(img)
       print("\nMinimum Filter")
       mini=min_filter(img)
       cv2_imshow(mini)
     elif ch==3:
       print("\nOriginal Image ")
       cv2_imshow(img)
       print("\nMaximum Filter")
       maxi=max_filter(img)
       cv2_imshow(maxi)
     elif ch==4:
       print("\nOriginal Image ")
       cv2_imshow(img)
       print("\nBilateral Filter")
       bi=bilateral(img)
       cv2_imshow(bi)
     elif ch==5:
      print('\nEnd')
      break
     else:
      print('Invalid Choice')
      continue
```

```python
main()
```
