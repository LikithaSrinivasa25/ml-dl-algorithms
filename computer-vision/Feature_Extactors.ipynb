```python
from google.colab.patches import cv2_imshow
import matplotlib.pyplot as plt
import cv2
import numpy as np
```

```python
def hog(image):
  from skimage.feature import hog
  from skimage import exposure
  ori = cv2.imread('/content/DOOOOGGGG.jpg')
  resized_image=cv2.resize(ori,(250,200))
  _, hog_image = hog(resized_image, orientations=20, pixels_per_cell=(16, 16),cells_per_block=(2, 2), visualize=True, multichannel=True)
  hog_image= exposure.rescale_intensity(hog_image,out_range=(0, 255))
  cv2_imshow(resized_image)
  cv2_imshow(hog_image)
```

```python
def orb(resized_image):
  orb = cv2.ORB_create(nfeatures=200)
  kp = orb.detect(resized_image, None)
  kp, des = orb.compute(resized_image, kp)
  img2 = cv2.drawKeypoints(resized_image, kp, None, color=(0, 255, 0), flags=0)
  cv2_imshow(resized_image)
  cv2_imshow(img2)
```

```python
def harris_corner(resized_image):
    resized_image=cv2.resize(resized_image,(300,300))
    resized_image1=cv2.resize(resized_image,(300,300))
    gray = cv2.cvtColor(resized_image1,cv2.COLOR_BGR2GRAY)
    gray = np.float32(gray)
    dst = cv2.cornerHarris(gray,2,3,0.04)
    dst = cv2.dilate(dst,None)
    resized_image[dst>0.01*dst.max()]=[0,0,255]
    cv2_imshow(resized_image1)
    cv2_imshow(resized_image)
```

```python
def sift(image,img):
  image=cv2.resize(image,(300,300))
  img1=cv2.resize(img,(300,300))
  gray = cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)
  sift = cv2.SIFT_create()
  kp, des = sift.detectAndCompute(gray,None)
  img=cv2.drawKeypoints(gray,kp,image,flags=cv2.DRAW_MATCHES_FLAGS_DRAW_RICH_KEYPOINTS)
  cv2_imshow(img1)
  cv2_imshow(image)
```

```python
def main():
  image=cv2.imread("/content/DOOOOGGGG.jpg")
  resized_image=cv2.resize(image,(300,300))
  while True:
    print("Feature Extactors \n1.Histogram of Feature Transformation \n2.Oriented FAST and rotated BRIEF (ORB) \n3.Harris Corner Dectection \n4.SIFT \n5.Exit")
    choice=int(input("Enter Choice = "))
    if choice==1:
      hog(resized_image)
    elif choice==2:
      orb(resized_image)
    elif choice==3:
      harris_corner(resized_image)
    elif choice==4:
      img=cv2.imread("/content/DOOOOGGGG.jpg")
      sift(image,img)
    elif choice==5:
      print("Thank you")
      break
    else:
      print("Invalid Choice")
```

```python
main()
```
