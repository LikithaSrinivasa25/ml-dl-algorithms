**UNIGRAM**

```python
from nltk.probability import FreqDist
import collections
text=input("Enter text=")
text=text.lower()
texts=text.split()
words=FreqDist(texts)
f=words.most_common()
print("The unigram:",f)
word=input("Please Enter the word that you need to find the unigram probability=")
f_word=words[word]
print(f"The number times {word} occurs in the text:",f_word)
print("The total number of words in the text=",len(texts))
unigram_probability=round(f_word/len(texts),4)
print(f"The probability of occurance of {word} in text is {unigram_probability}")
```

**BIGRAM**

```python
from nltk.tokenize import texttiling
from nltk import ngrams
from nltk.probability import FreqDist
from collections import Counter
text=input("Enter text =")
textt=text.split() #or else we can use word_tokenise(text)
bigram=ngrams(textt,2)
bigram_count=Counter(bigram)
bigram=bigram_count
print("The the bigram:",bigram)
total_bigram_count=bigram_count.values()
for bigram, count in bigram_count.items():
    bigram_probability=round(count/sum(total_bigram_count),4)
    if bigram_probability==0 or bigram_probability<=0.01:
      #Use laplace Smoothening Technique
      unique_vocabulary=set(textt)
      bigram_probability_smoothing=count+1/sum(total_bigram_count)+len(unique_vocabulary)
      print(f"The bigram {bigram} and its probabiltity using Laplace smoothening = {bigram_probability_smoothing}")
    else:
      print(f'Bigram {bigram} and its probability = {(bigram_probability)}')
```
