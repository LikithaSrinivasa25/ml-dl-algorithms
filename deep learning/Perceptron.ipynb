```python
import pandas as pd
import random
import numpy as np
```

```python
d=np.array([[1,0,1526,0],
        [0,1,1487,1],
        [0,0,1987,0],
        [1,1,6543,1],
        [1,0,7654,1],
        [0,0,7864,1],
        [1,1,2354,0],
        [0,1,8765,0],
        [1,0,2345,0],
        [1,0,6523,1]])
column=['Transaction_within_US','Transaction_within_24hrs','Transaction_amount($)','Fraudulent_Transaction']
```

```python
data=pd.DataFrame(data=d,columns=column)
```

```python
data
```

```python
X1=data["Transaction_within_US"]
X2=data["Transaction_within_24hrs"]
X3=data["Transaction_amount($)"]
Y=data["Fraudulent_Transaction"]
X0=1
```

```python
w0=random.randint(0,10)
print("The weight for bias is",w0)
w1=random.randint(0,10)
print(f"The weight for X1 is {w1}")
w2=random.randint(0,10)
print(f"The weight for X2 is {w2}")
w3=random.randint(0,10)
print(f"The weight for X3 is {w3}")
```

```python
def net_input(x0,x1,x2,x3,w0,w1,w2,w3):
    return (x0*w0)+(x1*w1)+(x2*w2)+(x3*w3)
```

```python
net=[]
for i in net_input(X0,X1,X2,X3,w0,w1,w2,w3):
    net.append(i)
```

```python
net
```

```python
def activation_func(net_ip):
    if net_ip>20000:
        return 1
    else:
        return 0
```

```python
y=list(Y)
act_func=[]
for i in net:
    x=activation_func(i)
    act_func.append(x)
print("The actual output is",y)
print("The predicted output is",act_func)
```

```python
j=0
for i in y:
    if act_func[j]==i:
        print(f"Actual output {i} || Predicted {act_func[j]} ----> no error")
    else:
        print(f"Actual output {i} || Predicted {act_func[j]} ----> error")
    j+=1
```
