Text Preprocessing

```python
def remove_punctuation(txt):
    punct_free_txt="".join([i for i in txt if i not in string.punctuation])
    print(punct_free_txt,"\n","--------------------------")
    return punct_free_txt
def remove_links(text):
    txt=re.sub("https\S+","",text)
    print(txt,"\n","--------------------------")
    return txt
def remove_emoji(text):
    no_emoji=emoji.replace_emoji(text)
    print(no_emoji,"\n","--------------------------")
    return no_emoji
def expand_contractions(text):
    contract=contractions.fix(text)
    print(contract,"\n","--------------------------")
    return contract
def replacing_numbers_to_words(text):
    p=inflect.engine()
    new_words=[]
    for word in text:
        if word.isdigit():
            new_word=p.number_to_words(word)
            new_words.append(new_word)
        else:
            new_words.append(word)
    print(new_words)
    return new_words
def tokenize(text):
    tokens=nltk.word_tokenize(text)
    print(tokens,"\n","--------------------------")
    return tokens
def remove_stopwords(text):
    new_words=[]
    for word in text:
        if word not in stopwords.words('english'):
            new_words.append(word)
    print(new_words,"\n","--------------------------")
    return new_words

print("Displaying Original Text: ")
print(txt,"\n","--------------------------")
print("Text without URLs: ")
no_url_text=remove_links(txt)
print("Text after handling contractions: ")
contract=expand_contractions(no_url_text)
print("Text without any punctuations: ")
no_puncts=remove_punctuation(contract)
print("Text without emoji: ")
no_emoji=remove_emoji(no_puncts)
#print("Text after converting numbers to words: \n")
#number_to_word=replacing_numbers_to_words(no_emoji)
print("Text after performing tokenization: \n")
tokens=tokenize(no_emoji)
print("Text after removing stopwords: \n")
stp_wrds=remove_stopwords(tokens)
```

Displaying the Bigrams and Trigrams in the corpus

```python
from nltk.util import ngrams

bi=list(nltk.bigrams(stp_wrds))
print("Bigrams Occuring in the Corpus: \n",bi)

tri=list(nltk.trigrams(stp_wrds))
print("\nTrigrams Occuring in the Corpus: \n",tri)

freq_bi = nltk.FreqDist(bi)
freq_tri = nltk.FreqDist(tri)
print ("\nMost common bigrams: ", freq_bi.most_common(5))
print ("\nMost common trigrams: ", freq_tri.most_common(5))
```

Prediction of a word given previous two and three words

```python
def prob(sentence,txt,pred_word):
  sent=txt.count(sentence)
  print("The total number of times <",sentence,"> occurs in the corpus is: ",sent)
  new_sent=sentence+" "+pred_word
  c=txt.count(new_sent)
  print("Total number of times <",sentence,"> is accompanied by <",pred_word,"> is: ",c)
  try:
    ans=c/sent
  except:
    if ZeroDivisionError:
      ans=0
  print("P (",pred_word,"|",sentence,") is:",round(ans,4))
  return ans
```

```python
print("Bigram Model")
print("Given two words: <widespread lack> prediction of 3rd word")
sentence="widespread lack"
sol=0
flg=0
for i in stp_wrds:
  pred_word=i
  a=prob(sentence,txt,pred_word)
  if a>sol:
    sol=a
    highest_prob=i
    flg+=1
  print("\n")
if flg!=0:
  print("Thus the word <",highest_prob,"> is most like to occur after <",sentence,"> as it has the highest probability of: ",round(sol,2))
else:
  print("Unable to predict the third word as the probability of getting any word in the corpus after <",sentence,"> is 0")
```

```python
print("Trigram Model")
print("Given three words: <widespread lack of> prediction of 3rd word")
sentence="widespread lack of"
sol=0
flg=0
for i in stp_wrds:
  pred_word=i
  a=prob(sentence,txt,pred_word)
  if a>sol:
    sol=a
    highest_prob=i
    flg+=1
  print("\n")
if flg!=0:
  print("Thus the word <",highest_prob,"> is most like to occur after <",sentence,"> as it has the highest probability of: ",round(sol,2))
else:
  print("Unable to predict the third word as the probability of getting any word in the corpus after <",sentence,"> is 0")
```
