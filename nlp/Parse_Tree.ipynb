```python
import nltk
from nltk.tree import*
from nltk.draw import tree
```

```python
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
```

```python
grammar = nltk.CFG.fromstring("""
S -> NP VP
PP -> P NP
NP -> Det N | Det N PP | 'I'
VP -> V NP | VP PP
Det -> 'an' | 'my'
N -> 'elephant' | 'pajamas'
V -> 'shot'
P -> 'in'
""")
sent = "I shot an elephant in my pajamas"
tokens = nltk.word_tokenize(sent)
parser = nltk.ChartParser(grammar)
trees = list(parser.parse(tokens))
```

```python
display(trees[0])
display(trees[1])
```

**Conclusion**: A parse tree or parsing tree or derivation tree or concrete syntax tree is an ordered, rooted tree that represents the syntactic structure of a string according to some context-free grammar. A parse tree was generated for the text "I shot an elephant in my pajamas".
