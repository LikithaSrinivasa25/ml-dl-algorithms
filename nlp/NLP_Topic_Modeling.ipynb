```python
from sklearn.decomposition import LatentDirichletAllocation
lda_model=LatentDirichletAllocation(n_components=20,learning_method='online',random_state=122,max_iter=100)
lda_top=lda_model.fit_transform(X)
```

```python
terms = vectorizer.get_feature_names_out()

for i, comp in enumerate(lda_model.components_):
    terms_comp = zip(terms, comp)
    sorted_terms = sorted(terms_comp, key= lambda x:x[1], reverse=True)[:7]
    print("Topic "+str(i)+": ")
    for t in sorted_terms:
        print(t[0],end=" ")
    print("\n")
```

```python
from sklearn.decomposition import NMF
```

```python
data=pd.read_csv("/content/drive/MyDrive/Colab Notebooks/abcnews-date-text.csv")
# This contains data of news headlines published over a period of nineteen years.
# Sourced from the reputable Australian news source ABC (Australian Broadcasting
# Corporation)
# Agency Site: (http://www.abc.net.au)
# Content
# publish_date: Date of publishing for the article in yyyyMMdd format
# headline_text: Text of the headline in Ascii , English , lowercase
# Start Date: 2003-02-19 ; End Date: 2021-12-31
```

```python
data_text=data[:5000][['headline_text']]
```

```python
documents=data_text
```

```python
documents.head()
```

```python
documents.isnull().sum()
```

```python
corpus=[]
for i in range(0,len(documents)):
    text=re.sub('[^a-zA-z]',' ',documents['headline_text'][i])
    text=text.lower()
    text=text.split()
    #text=[PorterStemmer().stem(word) for word in text if not word in stopwords.words('english')]
    text=' '.join(text)
    corpus.append(text)
```

```python
processed_docs1
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(stop_words='english',
max_features= 1000, # keep top 1000 terms
max_df = 0.5,
smooth_idf=True)

X = vectorizer.fit_transform(processed_docs1)

X.shape # check shape of the document-term matrix
```

```python
# Create an NMF instance: model
# the 10 components will be the topics
model = NMF(n_components=20, random_state=122,max_iter=100)

# Fit the model to TF-IDF
model.fit(X)

# Transform the TF-IDF: nmf_features
nmf_features = model.transform(X)
```

```python
terms = vectorizer.get_feature_names_out()

for i, comp in enumerate(model.components_):
    terms_comp = zip(terms, comp)
    sorted_terms = sorted(terms_comp, key= lambda x:x[1], reverse=True)[:7]
    print("Topic "+str(i)+": ")
    for t in sorted_terms:
        print(t[0],end=" ")
    print("\n")
```

**Conclusion**: Topic modeling is the method of extracting needed attributes from a bag of words. This is critical because each word in the corpus is treated as a feature in NLP. As a result, feature reduction allows us to focus on the relevant material rather than wasting time sifting through all of the data's text. Latent Dirichlet Allocation (LDA) is a generative statistical model that explains a set of observations through unobserved groups, and each group explains why some parts of the data are similar. Latent semantic analysis (LSA) is a technique in natural language processing, in particular distributional semantics, of analyzing relationships between a set of documents and the terms they contain by producing a set of concepts related to the documents and terms. Non-Negative Matrix Factorization is a statistical method that helps us to reduce the dimension of the input corpora or corpora. Internally, it uses the factor analysis method to give comparatively less weightage to the words that are having less coherence.
