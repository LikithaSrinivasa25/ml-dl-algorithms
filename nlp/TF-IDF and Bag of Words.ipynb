Bag of Words

```python
text1='Stranger Things is an amazing tv series!'
text2='Stranger Things is the best tv series!'
text3='Stranger Things is so great'
```

```python
l=[]
for i in [text1,text2,text3]:
  l.append(i)
print(l)
```

```python
class bow:
  def __init__(self,text):
    self.ltext=text.lower()
    self.text=contractions.fix(self.ltext)
    self.stop_words=set(adv.stopwords['english'])

  def remove_quotes_brackets(self):
    self.text1=re.sub("[\"\'\(\)]", "", self.text)
    return self.text1
  
  def remove_punctuations(self,text):
    text1=re.sub(r"[^\w\s]", " ",text)
    return text1 
  
  def tokenize(self,text):
    self.t_text=word_tokenize(text)
    return self.t_text
  
  def stop_word(self,text):
    self.sw=[]
    for i in text:
      if i not in self.stop_words:
        self.sw.append(i)
    return self.sw

  def count(self,l):
    d={}
    for i in l:
      d[i]=l.count(i)
    return d
  
  def preprocess(self):
    q_text=self.remove_quotes_brackets()
    p_text=self.remove_punctuations(q_text)
    t_text=self.tokenize(p_text)
    swtext=self.stop_word(t_text)
    word=' '.join(swtext)
    ctext=self.count(swtext)
    return t_text,swtext,ctext,word
```

```python
for i in [text1,text2,text3]:
  print()
  t=bow(i)
  tkn,stp,wrd,cnt=t.preprocess()
  print("\nTokenized text",tkn)
  print("\nText after Stopwords removal:",stp)
  print("\nCount of Words:",wrd)
```

TF-IDF

```python
doc=["Bob is clever than the others",
     "John loves to explore the stars at night",
     "The sun is shining in the sky",
     "The night sky is full of stars",]
```

```python
doc
```

```python
cv=CountVectorizer()
word_count_vector=cv.fit_transform(doc)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf_vectoriser=TfidfVectorizer(use_idf=True)
tfidf_vectoriser_vectors=tfidf_vectoriser.fit_transform(doc)
```

```python
f=tfidf_vectoriser.get_feature_names_out()
dense=tfidf_vectoriser_vectors.todense()
```

```python
tf_df=pd.DataFrame(dense.tolist(),columns=tfidf_vectoriser.get_feature_names_out())
tf_df
```
