```python
import pandas as pd
import gensim
```

```python
df=pd.read_csv("Amazon_Unlocked_Mobile1.csv", error_bad_lines=False)
df.head()
```

```python
df.duplicated()
```

```python
data=df.drop(["Product Name","Brand Name","Price","Rating","Review Votes"],axis=1)
data.head()
```

```python
data.Reviews[0]
```

```python
data.dropna(subset = ["Reviews"], inplace=True)
```

```python
reviewtxt=data["Reviews"].apply(gensim.utils.simple_preprocess)
reviewtxt
```

```python
model=gensim.models.Word2Vec(window=10,min_count=2)
```

```python
model.build_vocab(reviewtxt)
```

```python
model.epochs
```

```python
model.corpus_count
```

```python
model.train(reviewtxt,total_examples=model.corpus_count,epochs=model.epochs)
```

```python
model.wv.most_similar('bad')
```

```python
model.wv.similarity(w1="pissed",w2="good")
```

```python
model.wv.similarity(w1="lucky",w2="phone")
```

```python
model.wv.similarity(w1="low",w2="great")
```
