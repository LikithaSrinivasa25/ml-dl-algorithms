```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
text = "One antiplatelet agent is aspirin. Almost everyone with coronary artery disease, including those who have had a heart attack, stent, or CABG are treated with aspirin for the rest of their lives. A second type of antiplatelet agent, called a P2Y12 inhibitor, is usually prescribed for months or years in addition to the aspirin therapy."
doc = nlp(text)
print(nlp.pipe_names)
```

```python
from spacy import displacy
displacy.render(doc, style="ent", jupyter=True)
```

```python
text2='''Isaac and I communicated almost exclusively through sighs. In fact, on the Wednesday I made the acquaintance of Augustus Waters'''
text2
```
