```python
import pandas as pd
import numpy as np
import re
import string
from nltk.tokenize import word_tokenize
import csv
import seaborn as sns
import unicodedata
from nltk.stem import WordNetLemmatizer
import nltk
nltk.download('wordnet')
nltk.download('omw-1.4')
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv("/content/spam.csv",encoding="ISO-8859-1")
df.head()
```

```python
df["Message"]=df["v2"]
df["Classification"]=df["v1"]
```

```python
df.drop_duplicates()
```

```python
df["Classification"]=df["Classification"].map({"ham":0,"spam":1})
df.head(10)
```

```python
df["Message"] = df['Message'].replace(np.nan, '')
```

```python
def remove_url(text):
  url=re.compile(r'http?://\S+|www\.\S+')
  return url.sub(r'',text)
```

```python
df['Message']=df['Message'].apply(lambda x:remove_url(x))
```

```python
def remove_html(text):
  html=re.compile(r'<.*?>')
  return html.sub(r'',text)
```

```python
df["Message"]=df["Message"].apply(lambda x :remove_html(x))
```

```python
def remove_emoji(text):
  emoji=re.compile("["
                           u"\U0001F600-\U0001F64F"
                           u"\U0001F300-\U0001F5FF"
                           u"\U0001F680-\U0001F6FF"
                           u"\U0001F1E0-\U0001F1FF"
                           u"\U00002702-\U000027B0"
                           u"\U000024C2-\U0001F251"
                           "]+", flags=re.UNICODE)
  return emoji.sub(r'',text)
```

```python
df["Message"]=df["Message"].apply(lambda x: remove_emoji(x))
```

```python
def remove_puntucations(text):
  table=str.maketrans("","",string.punctuation)
  return text.translate(table)
```

```python
df['Message']=df['Message'].apply(lambda x: remove_puntucations(x))
```

```python
def lower(text):
    lower=text.lower()
    return lower
```

```python
df["Message"]=df["Message"].apply( lambda x: lower(x))
```

```python
def accent(text):
  text3=unicodedata.normalize('NFKD', text).encode('ascii', 'ignore').decode('utf-8', 'ignore')
  return text3
```

```python
df["Message"]=df["Message"].apply(lambda x: accent(x))
```

```python
def lemmization(text):
  lemmatizer = WordNetLemmatizer()
  text=lemmatizer.lemmatize(text)

  return (text)
```

```python
df["Message"]=df["Message"].apply(lambda x: lemmization(x))
```

```python
df["Message"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(df['Message'],df['Classification'],random_state=0)
```

```python
vect = CountVectorizer().fit(X_train)
X_train_vectorized = vect.transform(X_train)
clfrNB = MultinomialNB(alpha = 0.1)
clfrNB.fit(X_train_vectorized, y_train)
preds = clfrNB.predict(vect.transform(X_test))
score = accuracy_score(y_test, preds)
```

```python
print("Navie bayes Accuracy:",(score*100))
```
