```python
#Importing the data set
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

data=pd.read_csv("Melbourne_housing_FULL.csv")
display(data)
```

```python
# View first few records
data.head()
```

```python
# View data types
data.info()
```

```python
#Dropping the Categorial values
data= data.drop(['Suburb', 'Address', 'Type', 'Method', 'SellerG', 'Date', 'CouncilArea',
       'Regionname'],1)
```

```python
# Checking the null values in the dataset 

data.isna().sum()
```

```python
#Filling missing values with mean:
import math
data['Bedroom2']=data['Bedroom2'].fillna(data['Bedroom2'].mean())
```

```python
data.isna().sum()
```

```python
#Dropping the na values
data= data.dropna()
```

```python
 #Identify object columns
print(data.select_dtypes(['object']).columns)
```

```python
#Dimension of data set after droppping the categorical values
data.shape
```

```python
# Examine Rooms v Bedroom2
data['Rooms v Bedroom2'] = data['Rooms'] - data['Bedroom2']
data
```

```python
# As the difference between Rooms and Bedroom2 is minimal we can drop the bedrooms column, we can get better analysis
data = data.drop(['Bedroom2','Rooms v Bedroom2'],1)
```

```python
#Correlation
plt.figure(figsize=(10,6))
sns.heatmap(data.corr(),annot= True)
plt.title('Variable Correlation')
```

```python
#Linear Regression with Outliers

# Setting the dependent and independent features
x=data[['Rooms', 'Distance', 'Bathroom','Car', 'Landsize', 
            'BuildingArea', 'Propertycount']]
y = data['Price']
```

```python
# Train, test, split
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x,y, test_size = .20, random_state= 0)
```

```python
# Traing the data

from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
regressor.fit(x_train,y_train)
```

```python
# Predicting test results

y_pred = regressor.predict(x_test)
```

```python
# Evaluating the Accuracy of the model

from sklearn.metrics import r2_score
Accuracy=r2_score(y_test,y_pred)*100
print(" Accuracy of the model is %.2f" %Accuracy)
print(regressor.intercept_)
print(regressor.coef_)
```

```python
# Linear Regression without Outliers

# Checking for outliers in the 'Price' attribute which will be chosen for target variable
def outliers(data):
    sns.boxplot(data)
    plt.show()
    import math
    q1 = data.quantile(0.25)
    q3 = data.quantile(0.75)
    iqr = q3 - q1
    upper = math.floor(q3 + 1.5*iqr)
    lower = math.floor(q1 - 1.5*iqr)
    return (upper,lower)
```

```python
# Removal of outlier

upper_limit1=data['Price'].quantile(0.90)
lower_limit1=data['Price'].quantile(0.10)
```

```python
# Displaying outlier values

data[(data.Price < lower_limit1) | (data.Price > upper_limit1)]
```

```python
# Removal of outliers

data['Price']=np.where(data["Price"]>=upper_limit1,upper_limit1,data['Price'])
data['Price']=np.where(data["Price"]<=lower_limit1,lower_limit1,data['Price'])
```

```python
# Displaying the boxplot without outliers

sns.boxplot(data['Price'])
```

```python
#Now checking the Accuracy without outliers
# Setting the dependent and independent features

x=data[['Rooms', 'Distance', 'Bathroom','Car', 'Landsize', 
            'BuildingArea', 'Propertycount']]
y=data['Price']
```

```python
# Splitting the data into training and testing data

from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=0)
```

```python
# Traing the data

from sklearn.linear_model import LinearRegression
from sklearn import linear_model

linreg=LinearRegression()
linreg.fit(x_train,y_train)
```

```python
# Predicting test results

y_pred=linreg.predict(x_test)
y_pred
```

```python
# Evaluating the Accuracy of the model

from sklearn.metrics import r2_score
Accuracy=r2_score(y_test,y_pred)*100
print(" Accuracy of the model is %.2f" %Accuracy)
print(linreg.intercept_)
print(linreg.coef_)
```

```python
#Plotting for linear Regression model
plt.figure(figsize=(15,7))
sns.regplot(x=y_test,y=y_pred,ci=None,scatter_kws={"color": "green"}, line_kws={"color": "black"})
plt.show()
```
