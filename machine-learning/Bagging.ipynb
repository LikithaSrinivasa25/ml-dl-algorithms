```python
#Label encoding
from sklearn.preprocessing import LabelEncoder
from sklearn import preprocessing
label_encoder = preprocessing.LabelEncoder()
df['Sex']= label_encoder.fit_transform(df['Sex'])
df['ChestPainType']= label_encoder.fit_transform(df['ChestPainType'])
df['RestingECG']= label_encoder.fit_transform(df['RestingECG'])
df['ExerciseAngina']= label_encoder.fit_transform(df['ExerciseAngina'])
df['ST_Slope']= label_encoder.fit_transform(df['ST_Slope'])
df
```

```python
#onehot encoding
from sklearn.preprocessing import OneHotEncoder
enc=OneHotEncoder()
enc_df= pd.get_dummies(df, columns = ['Sex', 'ChestPainType', 'RestingECG', 'ExerciseAngina', 'ST_Slope'])
display(enc_df)
```

```python
#Splitting and training the variables
y = df['HeartDisease']
x = df.loc[:, df.columns != 'HeartDisease']
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.35, random_state=40)
print(x_train.shape); print(x_test.shape)
```

```python
#svm
from sklearn import svm
clf=svm.SVC(kernel='linear')
clf.fit(x_train,y_train)
y_pred=clf.predict(x_test)
```

```python
#Accuracy
from sklearn import metrics
print("Accuracy :",metrics.accuracy_score(y_test,y_pred))
```

```python
#Decision tree classifier model
from sklearn.tree import DecisionTreeClassifier
clf_gini = DecisionTreeClassifier(criterion='gini', max_depth=2, random_state=2)
clf_gini.fit(x_train, y_train)
```

```python
y_pred_gini = clf_gini.predict(x_test)
from sklearn.metrics import accuracy_score
print('Accuracy score with criterion gini index: {0:0.4f}'. format(accuracy_score(y_test, y_pred_gini)))
print('Training set score: {:.4f}'.format(clf_gini.score(x_train, y_train)))
print('Test set score: {:.4f}'.format(clf_gini.score(x_test, y_test)))
```

```python
#Plotting the decision tree
from sklearn import tree
plt.figure(figsize=(15,10))
tree.plot_tree(clf_gini.fit(x_train, y_train)) 
```

```python
#Bagging method
from sklearn.ensemble import BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
bagging=BaggingClassifier(DecisionTreeClassifier(),n_estimators=100,max_samples=0.8,oob_score=True,random_state=40)
bagging.fit(x_train,y_train)
```

```python
from sklearn.model_selection import cross_val_score
bagging_score=cross_val_score(bagging,x,y,cv=7)
bagging_score.mean()
```

```python
from sklearn.svm import SVC
bag_model_svc= BaggingClassifier(SVC(),n_estimators=33,oob_score=True,random_state=40)
bag_model_svc.fit(x_train,y_train)
```

```python
bag_model_svc.score(x_test,y_test)
```

```python
#plot for bagging - roc curve
from sklearn.metrics import plot_roc_curve
plot_roc_curve(bagging,x_test,y_test)
```

```python
#plot for decision tree
from sklearn.metrics import plot_roc_curve
plot_roc_curve(clf_gini,x_test,y_test) 
```

```python
#Covfusio matrix for Bagging
from sklearn.metrics import plot_confusion_matri
plot_confusion_matrix(bagging,x_test,y_test)
```

```python
# confusion matrix for decision tree
plot_confusion_matrix(clf_gini,x_test,y_test)
```
