**Program to implement various Image Enhancement Techniques**

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow
```

```python
image=cv2.imread('/content/image (1).jpg')
```

```python
def denoising():
  noise=cv2.imread('/content/noise.jpg')
  de=cv2.fastNlMeansDenoisingColored(noise,None,15,14,7,21)
  print("\nNoise-Image")
  cv2_imshow(noise)
  print('\nImager After Denoising')
  cv2_imshow(de)
```

```python
def sharpen():
  blur=cv2.imread('/content/blur.jpg')
  kernel = np.array([[0, -1, 0],
                   [-1, 5,-1],
                   [0, -1, 0]])
  sharp = cv2.filter2D(src=blur, ddepth=-1, kernel=kernel)
  print("\nBlur-Image")
  cv2_imshow(blur)
  print('\nImage After Sharpening')
  cv2_imshow(sharp)
  
```

```python
def blur():
  b=cv2.GaussianBlur(image,(9,9),0)
  print("\nOriginal-Image")
  cv2_imshow(image)
  print('\nBlur-Image')
  cv2_imshow(b)
```

```python
def detail():
  det=cv2.detailEnhance(image,sigma_s=15,sigma_r=0.2)
  print("\nOriginal-Image")
  cv2_imshow(image)
  print("\nDetail-Enhanced Image")
  cv2_imshow(det)
```

```python
def main():
  while True:
    print("\nImage Enhancement Techniques")
    print("1.Denoising")
    print("2.Sharpening")
    print("3.Blurring")
    print("4.Detail-Enhancement")
    print("5.Exit")
    ch=int(input("\nEnter your choice: "))
    if ch==1:
      print("\nImage Denoising")
      denoising()
    elif ch==2:
      print("\nImage Sharpening")
      sharpen()
    elif ch==3:
      print("\nImage Blurring")
      blur()
    elif ch==4:
      print("\nDetail-Enhancement")
      detail()
    elif ch==5:
      print("Thank You")
      break
    else:
      print("Invalid Choice")
      continue
main()
```
