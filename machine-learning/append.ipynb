```python
import pandas as pd
df=pd.read_csv("class1.csv")
print("Priniting Original Dataset:")
display(df)

a=list(df)
b=[]
values=pd.read_csv('class2.csv',header=None)
y=values.shape[0]
x=0
for i in range(y):
    d=list(values.iloc[x])
    c=zip(a,d)
    c_dict=dict(c)
    b.append(c_dict)
    x+=1
b1=pd.DataFrame(b)
print("Dataset to be appended in the original dataset: ")
display(b1)
df2=df.append(b,ignore_index=True)
print("New Dataset with appended values:")
display(df2)

```
