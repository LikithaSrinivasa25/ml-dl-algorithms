```python
# infer on a local image
pred=model(tab)
print(pred)
img = cv2.imread(tab)
cv2_imshow(img)
data = pred.xyxy[0].cpu().numpy()   # Get results as a numpy array
print(data)
crops = pred.crop(save=True)
```

```python
print(data)
ocr = PaddleOCR(use_angle_cls=True, lang='en')
```

```python
from dateutil import parser
from datetime import datetime
import re

# Define date patterns to search for in the text
date_patterns = [
  r"([A-Za-z]{3}\.\d{2,4})",    # Mon. YYYY or Month YYYY
  r"([A-Za-z]+\s?\d{2,4})",        # Month YYYY or Mon YYYY
  r"([A-Za-z]{3,}\s?\d{2})",       # Month YY or Mon YY
  r"(\d{1,2}/\d{2})",              # MM/YY
  r"(\d{1,2}[-/]\d{2})",           # MM-YY
  r"(\d{1,2}/\d{4})",              # MM/YYYY
]

# Define keywords for manufacturing date and expiry date
mdate_keywords = ['Mfg.Date', 'MFG', 'MFG.','MFG.DATE', 'Mfg.DATE', 'MFD', 'MFG.Dt','MFG.Dt.' 'MFG. DATE', 'mfd', 'mfg.Date', 'mfg.date']
edate_keywords = ['Expiry DATE', 'EXP', 'EXP.','EXPIRY DATE', 'EXPIRY', 'Expiry Date', 'EXP.DATE', 'EXP. DATE', 'EXP.Dt', 'EXP.Dt.','expiry date']


# Initialize variables
tablet_name = ""  # Initialize tablet name as an empty string
manufacturing_date = None
expiry_date = None

for detection in crops:
  crop_img = detection['im']
  modelResult = ocr.ocr(crop_img)
  print(modelResult)

  # Initialize a list to store lines of text
  text_lines = []

  # Assuming you want to perform OCR on the entire ROI
  # You can access the extracted text directly from modelResult
  for item in modelResult[0]:
      text = item[1][0]
      if text.strip():  # Check if the extracted text is not empty
          text_lines.append(text.strip())

  # Concatenate the text lines to form the complete tablet name
  if text_lines:
      tablet_name = " ".join(text_lines)

  # Search for manufacturing date and expiry date
  for line in text_lines:
      for pattern in date_patterns:
          matches = re.findall(pattern, line)
          if matches:
              for match in matches:
                  date_str = match.strip()
                  try:
                      date = parser.parse(date_str,fuzzy=True)
                      if any(keyword in line for keyword in mdate_keywords):
                          manufacturing_date = date
                      elif any(keyword in line for keyword in edate_keywords):
                          expiry_date = date
                  except ValueError:
                      pass

  # Calculate the number of days remaining
  if manufacturing_date and expiry_date:
      today = datetime.now()
      remaining_days = (expiry_date - today).days
      print("==============================")
      print("Tablet Name ---", tablet_name)
      print("Manufacturing Date:", manufacturing_date)
      print("Expiry Date:", expiry_date)
      print("Days Remaining to Expire:", remaining_days)
      print("==============================")
```
