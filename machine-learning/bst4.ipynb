```python
class Node:
    __slots__='element','left','right'
   
    def __init__(self, element, left=None, right=None):
        self.element=element
        self.left=left
        self.right=right
class BinaryTree(Node):
    def __init__(self):
        self.root=None
       
    def insert(self,troot,e):
        if troot:
            if e<troot.element:
                troot.left=self.insert(troot.left,e)
            elif e>troot.element:
                 troot.right=self.insert(troot.right,e)
        else:
            n=Node(e)
            troot=n
        return troot
       
    def preorder(self,troot):
        if troot:
            print(troot.element,end=" ")
            self.preorder(troot.left)
            self.preorder(troot.right)
           
    def inorder(self,troot):
        if troot:
            self.inorder(troot.left)
            print(troot.element,end=" ")
            self.inorder(troot.right)
           
    def postorder(self,troot):
        if troot:
            self.postorder(troot.left)
            self.postorder(troot.right)
            print(troot.element,end=" ")
    def height(self,r):
        if r is None:
            return 0
        return 1+max(self.height(r.left),self.height(r.right))
    def count(self,r):
        if r:
            x=self.count(r.left)
            y=self.count(r.right)
            return x+y+1
        return 0
           


def main():
    print("Creation of Binary Search Tree")
    s=BinaryTree()
    n=int(input("\nEnter the number of elements :"))
    e=[]
    for i in range(n):
        ele=int(input("Enter an element : "))
        e.append(ele)
    print(e)
    for i in e:
        s.root=s.insert(s.root,i)
    while True:
        print("\n1.Preorder")
        print("2.Inorder")
        print("3.Postorder")
        print("4.Height of tree")
        print("5.Count")
        print("6.Exit")
        ch=int(input("Enter your choice :"))
        if ch==1:
            print("\nPreorder traversal :")
            s.preorder(s.root)
            print()
        elif ch==2:
            print("\nInorder traversal :")
            s.inorder(s.root)
            print()
        elif ch==3:
            print("\nPostorder traversal :")
            s.postorder(s.root)
            print()
        elif ch==4:
            print("\nHeight of the tree :",s.height(s.root)-1)
        elif ch==5:
            print("\nCount of tree:",s.count(s.root))
        elif ch==6:
            print("End")
            break
```

```python
main()
```
