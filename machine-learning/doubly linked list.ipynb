```python
class Node:
    def __init__(self, element,prev,next):
        self.element=element
        self.next = next
        self.prev = prev

class DoublyLinkedList(Node):      

    def __init__ (self):
        self.header =Node(None,None,None)
        self.trailer =Node(None,None,None)
        self.header.next =self.trailer
        self.trailer.prev =self.header
        self.size = 0
   
    def __len__ (self) :
        return  self.size

    def  isempty(self) :
        if (self.size == 0) :
            return  True
        else :
            return  False
       
    def insert_between(self, e, predecessor,successor):
        newest =Node(e,predecessor,successor)
        predecessor.next =newest
        successor.prev =newest
        self.size+=1
        return newest
   
    def delete_node(self,node):
        predecessor = node.prev
        successor = node.next
        predecessor.next =successor
        successor.prev =predecessor
        self.size-=1
        ele=node.element
        node.element=None
        return ele

class DequeueLinked(DoublyLinkedList):
    def peekFront(self):
        if self.isempty():
            print("Dequeue is empty")
            return
        else:
            print(self.header.next.element)
            return

    def peekRear(self):
        if self.isempty():
            print("Dequeue is empty")
        else:
            print(self.trailer.prev.element)
            return

    def insert_first(self,e):
        self.insert_between(e,self.header,self.header.next)

    def insert_last(self,e):
        self.insert_between(e,self.trailer.prev, self.trailer)
       
   

    def delete_first(self):
        if self.isempty():
            print("Dequeue is empty")
        else:
            print("Element Deleted: ",self.delete_node(self.header.next))
       
           
    def delete_last(self):
        if self.isempty():
            print("Dequeue is empty")
        else:
            print("Element deleted: ",self.delete_node(self.trailer.prev))
    def display(self):
            if self.isempty():
                print("Dequeue is empty")
            else:
                temp =self.header
                while temp:
                    if(temp==self.header):
                        print(" ",temp.element,end = "<--")
                    elif(temp==self.trailer):
                        print(temp.element,end=" ")
                    else:
                        print(temp.element, end = "-->")
                    temp =temp.next
            print()
       
           
   
L = DequeueLinked()  
while(True):
    print("\n1. Insert at end")
    print("2. Insert at beginning")
    print("3. Delete element at Front")
    print("4. Delete element at Rear")
    print("5. Front element")
    print("6. Rear element")
    print("7. Is queue empty")
    print("8. Size")
    print("9. Display")
    print("10. Exit")
    choice=int(input("\nEnter your choice: "))
    if choice==1:
        element=input("Enter the element that you want to insert: ")
        print("Element Inserted: ",element)
        L.insert_last(element)
        L.display()
    elif choice==2:
        element=input("Enter the element that you want to insert: ")
        print("Element Inserted: ",element)
        L.insert_first(element)
        L.display()
    elif choice==3:
        if L.isempty()==True:
            print("queue is empty")
        else:
            L.delete_first()
            L.display()
    elif choice==4:
        if L.isempty()==True:
            print("queue is empty")
        else:
            L.delete_last()
            L.display()
    elif choice==5:
        L.peekFront()
        L.display()
    elif choice==6:
        L.peekRear()
        L.display()
    elif choice==7:
        print("Is Queue empty: ",L.isempty())    
    elif choice==8:
        print("Size: ",L.__len__())
        L.display()
       
    elif choice==9:
        if L.isempty():
            print("queue is empty")
        else:
            L.display()
    elif choice==10:
        print("Thank you")
        break
       
    else:
        print("Invalid input")
```
