```python
from dateutil import parser
from datetime import datetime
import re

name = ""
manufacture_date = None
expiry_date = None
tablet_name = ""  # Initialize tablet name as an empty string

for detection in crops:
    crop_img = detection['im']
    modelResult = ocr.ocr(crop_img)
    print(modelResult)

    m = []
    for res in range(0, len(modelResult[0])):
        text = modelResult[0][res][1][0]
        roi = modelResult[0][res][0]  # Region of Interest
        m.append(text)

    s = " ".join(m)
    p = s.replace("name0", "")

    if not tablet_name:
        # Check if the text matches a pattern of tablet name (starts with uppercase and no digits)
        if any(char.isalpha() for char in p):
            tablet_name = p.strip()

    if 'MFG' in p:
        mfg_parts = re.split(r'MFG.DATE|MFG|MANUFACTURE DATE', p, flags=re.IGNORECASE)
        for part in mfg_parts[1:]:
            try:
                manufacture_date = parser.parse(part.strip(), fuzzy=True)
                break  # Stop after finding one valid date format
            except ValueError:
                continue

    if 'EXPIRYDATE' in p:
        exp_parts = re.split(r'EXPIRYDATE|EXP|EXPIRY DATE', p, flags=re.IGNORECASE)
        for part in exp_parts[1:]:
            try:
                expiry_date = parser.parse(part.strip(), fuzzy=True)
                break  # Stop after finding one valid date format
            except ValueError:
                continue

print("\nTablet Name ---", tablet_name)
if manufacture_date is not None:
    print("Manufacture Date ---", manufacture_date)
if expiry_date is not None:
    print("Expiry Date ---", expiry_date)
    if datetime.now() < expiry_date:
        days_to_expire = (expiry_date - datetime.now()).days
        print("No of days remaining until Expiry ---", days_to_expire)
    else:
        print("The product has expired or Expiry Date is missing.")
```
