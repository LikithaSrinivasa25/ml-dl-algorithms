```python
import pandas as pd
df=pd.read_csv("data.csv")
display(df)
print("Current data type: ",df.dtypes)
print("\nNow modifying the data type..")
df['Date'] = pd.to_datetime(df['Date'])
print(df.dtypes)
print("\nModified data type: ",df.dtypes)
print("\nPrinting Date in a different format..")
df["Date"]=df['Date'].dt.strftime("%d-%b-%Y %H:%M:%p")
display(df)
```
