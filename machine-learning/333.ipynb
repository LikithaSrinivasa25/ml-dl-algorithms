```python
def percentage(sum):
    percentage=(sum/600)*100
    return(round(percentage,2))
def grade(percent):
    if(percent>=85):
        print("Grade:A \nResult:Distinction")
    elif(percent<85 and percent>=75):
        print("Grade:B \nResult:First class")
    elif(percent<75 and percent>=60):
        print("Grade:C \nResult:Second class")
    elif(percent<60 and percent>=40):
        print("Grade:D \nResult:Third class")
    else:
        print("Grade:F \nResult:Fail")
x='y'
while(x=='y' or x=='Y'):
    print("\nEnter marks of students:")
    m1=eval(input("Maths:"))
    m2=eval(input("Science:"))
    m3=eval(input("Social:"))
    m4=eval(input("Computer:"))
    m5=eval(input("Economics:"))
    m6=eval(input("arts:"))
    sum=m1+m2+m3+m4+m5+m6
    if(m1<35 or m2<35 or m3<35 or m4<35 or m5<35 or m6<35):
        print("Result:Fail")
    else:
        print("\nTotal marks:",sum)
        percent=percentage(sum)
        print("Percentage:",percent,"%")
        print(grade(percent))
    x=input("\n press'y' to continue else press anythyning else:")
print("Thank you!")
```

```python
str=input("Enter a string: ")
if len(str)==0:
    print("The string is empty")

else:
    char,digit,vowel,conso,space="","","","",0
    print("\nThe characters in the string are:",end="")
    for i in str:
        print(i,end=" ")
        if i.lower() in('a','e','i','o','u'):
            vowel+=i+" "
        elif i.isalpha():
            conso+=i+" "
        elif i.isdigit():
            digit+=i+" "
        elif i.isspace():
            space+=1
        else:
            char+=i+" "
    print("\nvowels in the string are:",vowel)
    print("consonants in the string are:",conso)
    print("digits in the string are:",digit) 
    print("special characters in the string are:",char)  
    print("number of spaces in the string are:",space)
```
