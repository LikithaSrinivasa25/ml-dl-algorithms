 **Resizing of the image**

```python
def re_size(): 
  height = img.shape[0] 
  width = img.shape[1] 
  n = int(input("How many times you want to resize do you want: ")) 
  for i in range(n): 
    scale_factor = float(input("Enter the scaleing factor: ")) 
    new_height = int(height * scale_factor) 
    new_width = int(width * scale_factor)
    dimensions = (new_width, new_height)
    new_image = cv2.resize(img, dimensions) 
    print("Original size of the image: ",img.shape)
    cv2_imshow(img) 
    print("New size of the image: ", new_image.shape) 
    cv2_imshow( new_image)
    new_image1 = cv2.resize(gray, dimensions)
    print("Original size of the image: ",gray.shape) 
    cv2_imshow(gray)
    print("New size of the image: ", new_image1.shape) 
    cv2_imshow( new_image1)
    new_image2 = cv2.resize(thresh, dimensions)
    print("Original size of the image: ",thresh.shape) 
    cv2_imshow(thresh)
    print("New size of the image: ", new_image2.shape)
    cv2_imshow( new_image2)
    print()
    print()
    
    
```
